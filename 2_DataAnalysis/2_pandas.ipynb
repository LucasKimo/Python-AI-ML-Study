# 8. 통계 함수

````python
print('합계:', df_copy['height'].sum()) # 합계
print('개수:', df_copy['height'].count()) # 개수, NaN은 포함하지 않음
print('평균:', df_copy['height'].mean()) # 평균
print('중앙값:', df_copy['height'].median()) # 중앙값
print('최대값:', df_copy['height'].max()) # 최대값
print('최소값:', df_copy['height'].min()) # 최소값
print('분산:', df_copy['height'].var()) # 분산
print('표준편차:', df_copy['height'].std()) # 표준편차
````


### **※ 평균과 중앙값**

#### 평균 (Mean)
- 데이터의 모든 값을 더한 뒤 데이터 개수로 나눈 값
- 전체 데이터의 중심적인 경향을 나타내는 대표적인 통계값
- [10, 20, 30]의 평균은 (10+20+30)/3 = 20
- 전체 데이터를 모두 반영하는 대표값
#### 중앙값 (Median)
- 데이터를 크기순으로 정렬했을 때 가장 가운에 위치한 값을 의미
- 이상치 (Outlier)의 영향을 적게 받는 특징이 있음
- [10, 20, 30]의 중앙값은 20. 
- 가운데 값이 두개: [10, 20, 30, 40]의 중앙값 = 20과 30의 평균값 >> 25.
- 데이터의 정중앙 위치를 나타내는 대표값

### **※ 분산과 표준편차**
#### 분산 (Variance)
- 데이터가 평균으로부터 얼마나 퍼져 있는지를 수치로 나타낸 값
- 데이터와 평균의 차이를 제곱한 뒤 평균을 내어 계산
- 값이 클수록 데이터의 흩어짐이 크다는 의미
- 값이 작을수록 데이터가 평균 근처에 있다는 의미
- 차이를 제곱해서 계산하기 때문에 단위가 원래 데이터와 달라짐.
#### 표준편차 (Standard Variation)
- 분산에 제곱근(sqrt)을 취한 값
- 원래 데이터와 같은 단위를 사용하므로 해석이 더 직관적.

# 13. get_dummies
- 머신러닝 모델은 숫자 데이터는 처리할 수 있지만 문자열 데이터는 직접 처리할 수 없다.
- 따라서 문자열로 이루어진 범주형 데이터를 숫자 형태로 변환하는 과정이 필요하며, 이때 많이 사용하는 방법이 원-핫 인코딩 (One-Hot Encoding)이다.
- Pandas에서는 pd.get_dummies() 함수를 이용해 쉽게 원-핫 인코딩을 수행할 수 있다.

### 원-핫 인코딩 (One-Hot Encoding)
- 범주형 데이터르르 각 항목별로 새로운 컬럼으로 분리하고, 해당 값이면 1, 아니면 을 부여하는 방법.
- 원-핫 인코딩은 범주형 데이터 (Categorical Data)에 적용한다.
- 범주형 데이터: 데이터의 값이 크기나 순서를 의미하는 것이 아니라 단순히 종류 (Category)를 나타내는 데이터를 말함.



- 문자열이라고 무조건 원-핫 인코딩하는 것은 아니다. 
  - (생년월일은 시간의 흐름이라는 의미가 있어서 안됨)
- 숫자열 데이터라고 해서 항상 수치형 데이터는 아니다. 
  - (등급 또는 학년 같은 값은 실제 숫자라기보다 특정 범주를 나타내는 경우가 많음. 이 경우 원-핫 인코딩 고려 가능) 


````python
df1_copy = df1.copy()
df1_copy
````

````python
blood_map = {'A':0, 'B':1, 'AB':2, 'O':3}
df1_copy['혈액형코드'] = df1_copy['혈액형'].map(blood_map)  # 라벨 인코딩
df1_copy.head()
````

````python
pd.get_dummies(df1_copy['혈액형'])  # 원-핫 인코딩
````

````python
df1_copy = pd.get_dummies(df1_copy, columns=['혈액형'])
df1_copy
````
