```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
shop = pd.read_csv('소상공인시장진흥공단_상가(상권)정보_서울_202603.csv', low_memory=False)
shop
```

```python
view_columns = ['상호명', '지점명', '상권업종대분류명', '상권업종중분류명',
                '상권업종소분류명', '시도명', '시군구명', '도로명', '도로명주소',
                '경도', '위도']

shop = shop[view_columns]
shop.head()
```
