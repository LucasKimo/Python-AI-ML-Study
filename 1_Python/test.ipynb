```python
logs = [
    "INFO: 사용자 로그인 성공",
    "DEBUG: 세션 생성 중",
    "ERROR: 데이터베이스 연결 실패",
    "INFO: 요청 처리 완료",
    "ERROR: 파일 저장 실패",
    "DEBUG: 응답 전송"
]

error_logs = []

for log in logs:
    # 로그가 ERROR로 시작하지 않으면 건너뜀
    if not log.startswith("ERROR"):
        continue
    # ERROR 로그만 리스트에 추가
    error_logs.append(log)

print("오류 로그만 출력:")
for e in error_logs:
  
  
    print(e)
```

```python
def func7(*args):
    return args  # 여기서 입력받은 모든 인자가 튜플 형태로 반환됩니다.

print(func7())            # ① 출력 결과: ()
print(func7(10))          # ② 출력 결과: (10,)
print(func7(10, 30, 50))  # ③ 출력 결과: (10, 30, 50)
```

```python
def func9(id, name, age):
    print(f'아이디: {id}')
    print(f'이름: {name}')
    print(f'나이: {age}')
    
func9('apple', '김사과', 20)
print()
func9(age=30, id='orange', name='오렌지')
print()

# 매개변수명과 딕셔너리의 키가 같아야함
# 딕셔너리의 키는 반드시 문자열 형태
dic1 = {'age':25, 'id':'banana', 'name':'반하나'}
func9(**dic1)
print()

# *의 데이터를 보낼 경우 키가 저장
func9(*dic1)
```

```python
def say_hello(name): # Callback 함수
    print(f"{name}님 안녕하세요!")

def execute(callback):
    print("작업을 시작합니다.")
    callback("김사과")
    print("작업을 종료합니다.")

execute(say_hello)
```

```python
def print_score(score):
    print(f"점수: {score}")

def print_pass_fail(score):
    if score >= 80:
        print(f"{score}점: 합격")
    else:
        print(f"{score}점: 불합격")

def print_bonus_score(score):
    print(f"보너스 적용 점수: {score + 5}")

def process_scores(scores, callback):
    for score in scores:
        callback(score)

scores = [80, 95, 70, 100, 60]

print("원점수 출력")
process_scores(scores, print_score)
print()

print("합격 여부 출력")
process_scores(scores, print_pass_fail)
print()

print("보너스 점수 출력")
process_scores(scores, print_bonus_score)
```

```python
class Fruit:
    def __init__(self, name, quantity, origin='원산지 미상'):
        self.name = name
        self.quantity = quantity
        self.origin = origin

# Fruit()  # 오류 발생
# TypeError: __init__() missing 2 required positional arguments

apple = Fruit('사과', 10, '한국')
print(apple)
print(apple.name)
print(apple.quantity)
print(apple.origin)

banana = Fruit('바나나', 5, '필리핀')
print(banana)
print(banana.name)
print(banana.quantity)
print(banana.origin)
```

```python
class Fruit:
    def __init__(self, name, quantity):
        self.name = name
        self.quantity = quantity

    # 인스턴스 메서드
    def print_info(self):
        print(f"과일 이름: {self.name}")
        print(f"수량: {self.quantity}")

    # 인스턴스 메서드
    def add_quantity(self, amount):
        self.quantity += amount
        print(f"{amount}개 추가되었습니다.")


apple = Fruit("사과", 10)
apple.print_info()
apple.add_quantity(5)
apple.print_info()
```
